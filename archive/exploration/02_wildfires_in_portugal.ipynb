```python
# General libraries
import pandas as pd
```

```python
# Population change

url_population = "https://ec.europa.eu/eurostat/api/dissemination/sdmx/3.0/data/dataflow/ESTAT/demo_r_gind3$defaultview/1.0?compress=false&format=csvdata&formatVersion=2.0&lang=en&labels=both"

df_population = pd.read_csv(url_population)

# Population density

url_density = "https://ec.europa.eu/eurostat/api/dissemination/sdmx/3.0/data/dataflow/ESTAT/demo_r_d3dens$defaultview/1.0?compress=false&format=csvdata&formatVersion=2.0&lang=en&labels=both"

df_density = pd.read_csv(url_density)

# Current population and age structure

url_current = "https://ec.europa.eu/eurostat/api/dissemination/sdmx/3.0/data/dataflow/ESTAT/demo_r_pjangrp3$defaultview/1.0?compress=false&format=csvdata&formatVersion=2.0&lang=en&labels=both"

df_current = pd.read_csv(url_current)
```

```python
print(df_population.columns.tolist())
print(df_density.columns.tolist())
print(df_current.columns.tolist())
```

```python
print(df_population.shape)
print(df_density.shape)
print(df_current.shape)
```

```python
# Keep just Portugal

geo_col ="geo: Geopolitical entity (reporting)"

df_population_pt = df_population[df_population[geo_col].str.startswith("PT")]

# Keep just Portugal
df_density_pt = df_density[df_density[geo_col].str.startswith("PT")]

# Keep just Portugal
df_current_pt = df_current[df_current[geo_col].str.startswith("PT")]
```

```python
print(df_current_pt.shape)
print(df_density_pt.shape)
print(df_current_pt.shape)
```

```python
# Check column names
print(df_population_pt.columns)
print(df_density_pt.columns)
print(df_current_pt.columns)
```

```python
# Check missing values
print(df_population_pt.isna().sum())
print(df_density_pt.isna().sum())
print(df_current_pt.isna().sum())
```

```python
# Check years available
print(df_population_pt["TIME_PERIOD: Time"].min(),
      df_population_pt["TIME_PERIOD: Time"].max())

print(df_density_pt["TIME_PERIOD: Time"].min(),
      df_density_pt["TIME_PERIOD: Time"].max())

print(df_current_pt["TIME_PERIOD: Time"].min(),
      df_current_pt["TIME_PERIOD: Time"].max())
```

```python
# Check which Portuguese regions are included
print(df_population_pt["geo: Geopolitical entity (reporting)"].unique())
```

```python
geo_col = "geo: Geopolitical entity (reporting)"

# Split "PT111: Alto Minho" into code and name
df_density_pt["geo_code"] = df_density_pt[geo_col].str.split(": ").str[0]
df_density_pt["geo_name"] = df_density_pt[geo_col].str.split(": ").str[1]

# Keep only NUTS3 (codes with 5 characters, like "PT111")
df_density_nuts3 = df_density_pt[df_density_pt["geo_code"].str.len() == 5]

print(df_density_nuts3["geo_code"].unique())
```

## Data

```python
# general libraries
from pathlib import Path
import pandas as pd
```

```python
# Load Data
import kagglehub

# Wildfire dataset

wildfire_path = kagglehub.dataset_download("fantineh/next-day-wildfire-spread", output_dir=Path("../data/wildfire/next-day-wildfire-spread"))

# Biodiversity dataset

biodiversity_path = kagglehub.dataset_download("nationalparkservice/park-biodiversity/versions/3", output_dir=Path("../data/biodiversity"))

print("Wildfire data:", wildfire_path)
print("Biodiversity data:", biodiversity_path)
```

```python
# Read biodiversity data

biodiversity_dir = Path("../data/biodiversity")

parks = pd.read_csv(biodiversity_dir / "parks.csv")
species = pd.read_csv(biodiversity_dir / "species.csv")

print(parks.shape)
print(species.shape)
```

```python
print(species.columns.tolist())
print(species.dtypes)
```

```python
# Info biodiversity data

parks.info()
species.info()
```

```python
# Read wildfire data

wildfire_dir = Path("../data/wildfire/next-day-wildfire-spread")
for file in wildfire_dir.iterdir():
    print(file.name, file.suffix)
```

```python
# sort

train_files = sorted(wildfire_dir.glob("*train_*.tfrecord"))
eval_filess = sorted(wildfire_dir.glob("*eval_*.tfrecord"))
test_files = sorted(wildfire_dir.glob("*test_*.tfrecord"))

print("Train:", len(train_files))
print("Eval:", len(eval_filess))
print("Test:", len(test_files))
```

Quick Sanity Test:

```python
species_parks = set(species["Park Name"].unique())
parks_table = set(parks["Park Name"].unique())

print("Parks in species:", len(species_parks))
print("Parks in parks.csv:", len(parks_table))

print("Missing from parks.csv:")
print(species_parks - parks_table)
```

```python
top_parks = (
    species.groupby("Park Name")["Species ID"]
    .nunique()
    .sort_values(ascending=False)
    .head(10)
)

top_parks
```

```python
top_10_names = top_parks.index

top_10_data = species[species["Park Name"].isin(top_10_names)]

print("Total species records:", len(top_10_data))
print("Unique species:", top_10_data["Species ID"].nunique())

coords_available = parks[
    parks["Park Name"].isin(top_10_names)
][["Latitude", "Longitude"]].notna().all(axis=1).sum()

print("Parks with coordinates:", coords_available, "/ 10")
```

```python
print(
    "Unique scientific species across top 10:",
    top_10_data["Scientific Name"].nunique()
)
```

## Purpose
Wildfire is among the most damaging recurrent hazards Portugal faces. ICNF has
recorded a median of roughly 21,000 rural ignitions a year since 2001, and in
2017 some 540,000 hectares burned — 6.1% of the mainland in a single season
[@icnf2025]. The conditions behind it are shifting: the rural interior is
losing population and ageing [@ine_aer], and farmland that is no longer worked
reverts to shrub. Whether that change in who lives on the land translates into
more fire is a question with direct budgetary consequences, because prevention
funding and land-maintenance measures are allocated geographically.

The mechanism is well argued in the literature and has been modelled forward in
simulation [@salis2022], but rarely tested backwards: against what actually
happened, at municipality level, across an entire country, using records the
state already keeps.

This report does that, for the Government of Portugal and from its own public
data. It has three parts:

1. **One reproducible panel.** ICNF fire statistics, INE demography, EFFIS burn
   perimeters and GADM boundaries joined at municipality-year level, rebuilt
   with a single command as new years are published.
2. **A statistical test.** Whether demographic ageing and rural depopulation are
   associated with fire incidence and with burned-area impact, holding
   municipality and year effects constant.
3. **A prioritisation tool.** A ranking of municipalities by risk of high
   burned-area impact, plus an interactive simulation for playing intervention
   scenarios through before budget is committed to them.

The report also states explicitly which decisions these data can support and
which would require records Portugal does not currently hold — so that the
boundary of the evidence is part of the deliverable, not a caveat discovered
later.

## Sustainable Development Goals framing
This project speaks primarily to **SDG 15 (Life on Land)**, target 15.3 on
combating desertification and restoring degraded land: across Mediterranean
agropastoral landscapes, farmland abandonment is an established driver of the
shrub regrowth that carries large fires [@salis2022]. Through fire it touches
**SDG 13 (Climate Action)** as Mediterranean fire regimes intensify under
warming, and **SDG 15.5**, since burned land is habitat lost.

## Why this matters for Portugal
The exposure is structural, not cyclical. Between 2019 and 2024, 38% of
mainland municipalities lost population — one of them by 16.6% — and the median
municipality now counts 247 elderly residents for every hundred children
[@ine_aer]. Whatever the relationship between that trend and fire turns out to
be, the conditions it describes will not reverse within the horizon of any
prevention plan.

The damage also runs wider than hectares. Burned land is habitat lost under
**SDG 15.5** [@owid_sdg15], and smoke is a public-health exposure: wildfires are
one of Portugal's most variable sources of fine particulate matter (PM2.5)
[@owid_pm25].

Prevention, meanwhile, is a finite budget spread across 278 mainland
municipalities. Any improvement in where it is directed is a direct return —
which is why Chapter 04 aims at a ranking rather than an explanation.

## Anchor study
The closest precedent for this project is Salis et al. [@salis2022], who asked the same question in a Mediterranean agropastoral landscape in north-western Sardinia: what happens to fire when the people who managed the land stop managing it?
Their design is the mirror image of ours. They simulated ~17,000 fire seasons under nine land-abandonment scenarios (0/10/20/30/40% of agropastoral land, two spatial patterns) and measured the effect on burn probability, flame length and fire size. Abandonment of 40% of agropastoral land raised average annual burned area from about 2,400 ha to about 3,100 ha, and roughly doubled the share of the landscape in the high-exposure class. Crucially, they derived which of their 26 municipalities were most affected — turning a mechanism into a prioritisation.
Two details make the paper a tight anchor rather than a loose reference:
1. **They needed a demographic proxy for abandonment and built one.** Their municipal abandonment index combines agro-silvopastoral working units with the change in municipal population (2021 vs. 2001). That is close to our Part A predictor, though theirs also carries an
agricultural labour component this project does not hold.
2. **They state the gap we fill.** Their abandonment percentages are assumed, not observed; their landscape is 120,000 ha of one Italian island; and the effect is simulated forward rather than estimated from what actually happened.
What this project adds. We test the same mechanism in the observational direction, on a municipality-year panel covering the mainland municipalities, with demography available for 2019–2024. Where Salis et al. ask what would happen if X% were abandoned, we ask whether the demographic conditions that drive abandonment are visible in what has actually burned since. If the two converge, the simulated effect sizes gain empirical support outside Sardinia; if they diverge, the difference is itself informative about how far fire-spread simulations travel across Mediterranean contexts.

## Hypothesis and research questions
**Hypothesis:** there is potential to reduce wildfire incidence by maintaining
abandoned meadows and agricultural land — depopulation and ageing lead to land
abandonment, which increases fuel load and fire risk.

This chain has three links. The project measures the first directly
(demography, INE) and the last directly (fire, ICNF and EFFIS). The middle link
— land abandonment itself — is not observable in the data available, so it is
carried as an assumption supported by the literature rather than tested here.

**Research questions:**
- **A (Chapter 03):** Are demographic ageing and rural depopulation associated
  with wildfire incidence and with burned-area impact at the municipality level?
- **B (Chapter 04):** Can municipality-level demographic and settlement
  conditions rank municipalities by their risk of high burned-area impact?


## Data sources
Four public sources, joined at municipality-year level (Portugal's 278
mainland *concelhos*, 2001–2025 for fire history, 2019–2024 for demography):

- **EFFIS** (Copernicus) — satellite-mapped burn perimeters
- **ICNF** — official rural fire statistics, causes and severity
- **INE** — municipality-level demography (ageing, density, growth)
- **GADM** — administrative boundaries, used to join the above by municipality code

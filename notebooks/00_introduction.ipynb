## Sustainable Development Goals framing

This project speaks primarily to **SDG 15 (Life on Land)**, specifically target
15.3 on combating desertification and restoring degraded land — rural land
abandonment in Portugal is a documented driver of the shrub and woodland
regrowth that fuels large fires. It also connects to **SDG 13 (Climate Action)**,
since wildfire regimes are intensifying under climate change, and to
**SDG 11 (Sustainable Cities and Communities)**, given the resilience question
for rural municipalities losing population.


## Why Portugal

Portugal combines three things this project needs: (1) among the highest
wildfire incidence in Europe, with catastrophic years like 2017, (2) one of
the EU's fastest-aging rural populations, concentrated in exactly the
interior/northern regions most affected by fire, and (3) unusually good public
data — EFFIS, ICNF and INE together allow a municipality-year panel spanning
two decades, which most countries cannot offer.


## Anchor study

The closest precedent for this project is Salis et al. [@salis2022], who asked the same question in a Mediterranean agropastoral landscape in north-western Sardinia: what happens to fire when the people who managed the land stop managing it?
Their design is the mirror image of ours. They simulated ~17,000 fire seasons under nine land-abandonment scenarios (0/10/20/30/40% of agropastoral land, two spatial patterns) and measured the effect on burn probability, flame length and fire size. Abandonment of 40% of agropastoral land raised average annual burned area from about 2,400 ha to about 3,100 ha, and roughly doubled the share of the landscape in the high-exposure class. Crucially, they derived which of their 26 municipalities were most affected — turning a mechanism into a prioritisation.
Two details make the paper a tight anchor rather than a loose reference:
1. **They needed a demographic proxy for abandonment and built one.** Their municipal abandonment index combines agro-silvopastoral working units with the change in municipal population (2021 vs. 2001). That is, almost exactly, our Part A predictor — used by them as a scenario input, by us as a measured variable.
2. **They state the gap we fill.** Their abandonment percentages are assumed, not observed; their landscape is 120,000 ha of one Italian island; and the effect is simulated forward rather than estimated from what actually happened.
What this project adds. We test the same mechanism in the observational direction, on a municipality-year panel covering all of mainland Portugal over two decades. Where Salis et al. ask what would happen if X% were abandoned, we ask where abandonment has already happened, and what has burned there since. If the two converge, the simulated effect sizes gain empirical support outside Sardinia; if they diverge, the difference is itself informative about how far fire-spread simulations travel across Mediterranean contexts.


## Hypothesis and research questions

**Hypothesis:** there is potential to reduce wildfire incidence by maintaining
abandoned meadows and agricultural land — depopulation and aging lead to land
abandonment, which increases fuel load and fire risk.

**Research questions:**
- **A (Chapter 03):** Is demographic aging and rural depopulation associated
  with wildfire incidence and severity at the municipality level?
- **B (Chapter 04):** Can municipality-level demographic and land conditions
  predict where fires occur?


## Data sources

Four public sources, joined at municipality-year level (Portugal's 278
mainland *concelhos*, 2001–2025 for fire history, 2019–2024 for demography):

- **EFFIS** (Copernicus) — satellite-mapped burn perimeters
- **ICNF** — official rural fire statistics, causes and severity
- **INE** — municipality-level demography (aging, density, growth)
- **GADM** — administrative boundaries, used to join the above by municipality code

## What we found
This report set out to test whether demographic change — an ageing population and rural depopulation — is associated with wildfire at the level of Portuguese municipalities, and whether any such association is strong enough to guide where prevention effort is directed. Four findings answer that, and they are best read in order, because each qualifies the one before it.

**The association between ageing and fire is real but weak, and it sits mostly between regions rather than within them.** Measured across all 278 mainland municipalities, the ageing index and the share of municipal area burned correlate at ρ ≈ 0.10 — positive, and small. That figure does not survive disaggregation: inside the north it falls to ρ ≈ 0.02, and inside the south it reverses to ρ ≈ −0.24, where older municipalities burn less. The north is both older and burnier than the south, and that single fact produces most of the national number (Chapter 02). Read without the split, the same correlation would appear to support a causal story the geography does not.

**Fire frequency and fire impact answer differently.** With municipality and year effects held constant, demographic structure shows no clear link to how often fires start, and a weak positive association with how much land burns — a coefficient of 0.669 on the elderly share, marginally significant at p = 0.038, whose significance weakens once population change enters the model, and which lifts explained variance only from 0.205 to 0.230 (Chapter 03). The distinction matters for policy: the two outcomes are not interchangeable, and the demographic signal attaches to severity, not to occurrence. It is also why burned-area ratio, rather than the number of fires, is the outcome measured throughout this report — a municipality can record few fires in a year that devastates it (Chapter 03).

**A weak effect still leaves a usable ranking.** The same demographic and settlement variables, used predictively rather than explanatorily, identify high burned-area impact at an average precision of 0.409 against a 19.1% base rate, with eight of the twenty highest-ranked municipality-years actually high-impact — and twelve not (Chapter 04). Explaining little and ranking usefully are different tasks, and this project succeeds at the second more than at the first.

**Fire years are not predictable from fire history.** No fitted time-series model beat the long-run mean of the national burned-area series, and 2025 fell far outside anything that history implied (Chapter 03). Whatever drives a catastrophic season, it is not momentum in the series itself.


## Answering the research questions
**Question A — partly.** For burned-area impact the answer is a qualified yes: the association is positive, weak, and marginally significant. For fire incidence it is no. And the raw national correlation must not be quoted as the answer, since it is largely compositional rather than a within-municipality relationship.

**Question B — yes, usefully.** Demographic and settlement conditions rank municipalities roughly twice as well as chance at identifying high burned-area impact. The output is a relative priority ordering, not a calibrated probability of future fire, and it should be read and used as one.

**The hypothesis itself is neither confirmed nor refuted.** It claims that maintaining abandoned meadows and agricultural land would reduce wildfire. This project measures the first link of that chain, demography, and the last, fire, and finds them weakly consistent with it. The middle link — abandonment itself — was never measured, so the mechanism remains an assumption carried from the literature. The direction of our results does not contradict the effects simulated for Sardinia; it cannot confirm their magnitude.

## Limitations
**The middle link is unmeasured.** The only land-cover variable available describes the composition of burn scars, not of municipalities: it exists only where fire already occurred, and it is uncorrelated with ageing (ρ ≈ −0.02). Every statement here about abandonment as a mechanism is inference, not evidence.

**Municipal land cover is absent.** Answering "do abandoned municipalities burn more?" requires wall-to-wall land cover per municipality, which this project does not hold. Population density is not a substitute — it correlates with burn rate about as weakly as ageing does.

**Six years of demography is little variation to exploit.** INE publishes municipal demography for 2019–2024 only, and ageing moves slowly. The limited evidence in Chapter 03 is equally consistent with a weak true effect and with a real effect that this window is too short to detect.

**Weather is absent by design.** Drought, wind and temperature are not in the panel, so no coefficient reported here is a complete account of fire risk.

**Measurement breaks constrain comparison.** ICNF changed its burned-area convention in 2017 with no overlap year, and EFFIS lowered its detection threshold around 2020 (Chapter 01). Neither series can be read across those breaks without explicit adjustment.

## Policy implications
**Use the ranking as triage, not as forecast.** It narrows 278 mainland municipalities to a shortlist of roughly twenty for closer assessment. That is a meaningful reduction in search space; it is not a prediction of where fire will occur.

**Do not allocate on ageing alone.** A nationwide rule equating old with at-risk would misdirect resources: south of the Tejo the relationship runs the other way. Demographic structure is a weak national signal assembled from opposite regional ones.

**Do not read falling ignition counts as success.** Ignitions have declined since 2001 while long-duration and very large fires have not (Chapter 02). Four fifths of all fires stay under one hectare, while the 0.2% above 500 hectares produce the burned-area totals — a prevention effort measured in ignitions counts the tail that does not matter. Fewer fires is not less fire.

**The decisive data gap is one the client can close.** Municipal land cover, published nationally as COS, would make the abandonment link measurable rather than assumed — and would allow the central question to be answered directly instead of approached from both ends.

## Conclusion
Demographic change is a real but modest part of Portugal's wildfire picture, and not a usable explanation on its own. Where it helps is narrower than the hypothesis assumed: it says little about where fires start, something about how much land burns, and enough — in combination — to rank municipalities for attention. The mechanism the hypothesis rests on could not be tested with the data available, and closing that gap is a decision for the client rather than a limitation of method. Until then, the defensible use of this work is prioritisation, not prediction.
